Source Text

Prompt engineering is the process of designing instructions that help an AI system produce useful results. Effective prompts often include context, a clear task, constraints, and a desired output format. Prompt engineering techniques include prompt chaining, ReACT prompting, and self-reflection. Prompt chaining breaks a larger task into smaller connected prompts. ReACT combines planning with actions and observations. Self-reflection asks AI to evaluate its own response and revise weaknesses. Effective prompts are often tested and improved based on the quality of their outputs.

Original Summary :

Common techniques include prompt chaining, which divides tasks into smaller steps; ReACT prompting, which combines planning, actions, and observations; and self-reflection, where AI reviews and improves its own response. Prompts are often tested and revised to improve the quality of the final output.

AI Self Critique : - Prompt engineering means writing clear instructions that help an AI system produce useful and accurate results. Strong prompts usually include context, a clear task, limits, and the desired response format.

- Prompt chaining breaks a large task into smaller connected steps. ReACT prompting has the AI plan, take actions, and use the results to decide what to do next.

- Self-reflection asks the AI to review its own response, identify weaknesses, and improve it. Prompts are often tested and revised to produce better results.

```python
original_summary = """
Prompt engineering is the process of creating clear instructions to help AI produce useful results. Good prompts usually include context, a specific task, constraints, and an output format. Common techniques include prompt chaining, which divides tasks into smaller steps; ReACT prompting, which combines planning, actions, and observations; and self-reflection, where AI reviews and improves its own response. Prompts are often tested and revised to improve the quality of the final output.
"""

improved_summary = """
- Prompt engineering means creating clear instructions that help an AI system produce useful results. Good prompts usually include context, a clear task, limits, and the desired response format.

- Prompt chaining breaks a large task into smaller connected steps. ReACT prompting allows the AI to plan, take an action, observe the result, and decide what to do next.

- Self-reflection asks the AI to review its own response, find weaknesses, and improve it. Prompts are often tested and revised based on the quality of the results.
"""

print("SELF-REFLECTION COMPARISON")
print("--------------------------------")

print("\nORIGINAL SUMMARY:")
print(original_summary.strip())
print("\nOriginal word count:", len(original_summary.split()))

print("\nIMPROVED SUMMARY:")
print(improved_summary.strip())
print("\nImproved word count:", len(improved_summary.split()))
```
