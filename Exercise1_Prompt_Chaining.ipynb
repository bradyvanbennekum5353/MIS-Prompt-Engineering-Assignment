Exercise 1 — Prompt Chaining for a Customer Support AI

Tools Used:
- ChatGPT
- Google Colab
- Python

Goal:
Create a multi-step customer-support prompt chain where the output of one prompt is used as input for the next prompt.

Customer Message:
"I ordered a blue hoodie, but I received a red T-shirt. I want the correct item."

 Step 1 — Classify the Customer Issue

Prompt:

You are a customer service AI for an online clothing store.

Read the customer's message and classify the problem into exactly one of these categories:

- Wrong Item
- Wrong Size
- Shipping Problem
- Refund Request
- Damaged Item
- Other

Return only the category name.

Customer message:
"I ordered a blue hoodie, but I received a red T-shirt. I want the correct item."

Wrong Item
Output:
Wrong Item

Step 2 — Gather Missing Information

Output received from Step 1:
Wrong Item

Prompt:

You are continuing a customer service case.

The customer's issue was classified as:
Wrong Item

Original customer message:
"I ordered a blue hoodie, but I received a red T-shirt. I want the correct item."

Determine what essential information is still missing before the problem can be solved.

Ask no more than two short questions.

Do not ask for information that the customer already provided.

1. What is your order number?
2. Can you provide a photo of the item you received?

Customer response:
"My order number is A12345, and yes, I can provide a photo."


Step 3 — Propose a Solution

Information from previous steps:

Issue classification:
Wrong Item

Customer message:
"I ordered a blue hoodie, but I received a red T-shirt. I want the correct item."

Additional information:
Order number A12345.
The customer can provide a photo.

Prompt:

You are a customer service AI for an online clothing store.

Using the information above, propose a simple solution.

Requirements:
- Be polite.
- Apologize briefly.
- Explain the next action.
- Do not promise anything outside normal store policy.
- Keep the response under 80 words.

I'm sorry you received the wrong item. Please upload a photo of the red T-shirt you received so the order can be verified. Once the order is confirmed, the support team can begin the process of replacing it with the blue hoodie you originally ordered.

Step 4 — Determine Whether to Escalate

Previous result:
The customer received the wrong product and requested a replacement.

Order number:
A12345

Prompt:

Determine whether this issue can be handled through the normal replacement process or should be escalated to a human support agent.

Escalate if:
- There is a payment dispute.
- The requested replacement is unavailable.
- The customer threatens legal action.
- The customer's identity or order cannot be verified.

Return:

Decision: Escalate or Do Not Escalate

Reason: One sentence explaining why.

Decision: Do Not Escalate

Reason: This is a standard wrong-item replacement request and none of the escalation conditions are present.

Testing and Iteration

Original Prompt:
"Tell me what is wrong with the customer's order."

Problem:
The prompt was too vague and did not control the format of the AI's answer.

Improved Prompt:
"Classify the customer's problem into exactly one category: Wrong Item, Wrong Size, Shipping Problem, Refund Request, Damaged Item, or Other. Return only the category name."

Improvement:
The revised prompt is more specific and produces a consistent output that can easily be passed to the next step in the prompt chain.

```python
customer_message = "I ordered a blue hoodie, but I received a red T-shirt. I want the correct item."

classification = "Wrong Item"

missing_information = [
    "Order number",
    "Photo of received item"
]

order_number = "A12345"

solution = """
Apologize to the customer, verify the incorrect item,
and begin the replacement process for the blue hoodie.
"""

escalation_decision = "Do Not Escalate"

print("CUSTOMER SUPPORT PROMPT CHAIN")
print("--------------------------------")
print("Customer Message:", customer_message)
print("Step 1 - Classification:", classification)
print("Step 2 - Missing Information:", missing_information)
print("Step 3 - Order Number:", order_number)
print("Step 3 - Proposed Solution:", solution.strip())
print("Step 4 - Escalation Decision:", escalation_decision)
```
