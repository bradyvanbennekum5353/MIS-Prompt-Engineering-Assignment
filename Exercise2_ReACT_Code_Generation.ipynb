Create a Python function called calculate_order_stats() that analyzes a list of customer order amounts.

Follow these stages:

Requirements:
- Function name must be calculate_order_stats().
- Input is a list of order amounts.
- Ignore negative values.
- Ignore non-numeric values.
- Return the number of valid orders.
- Return the average order amount rounded to two decimal places.
- Return the highest valid order amount.
- Return the lowest valid order amount.
- Handle an empty list without crashing.

STAGE 3 — RUN

STAGE 5 — FIX

STAGE 6 — FINAL
Provide the final corrected Python code ready to paste into Google Colab.

```python
def calculate_order_stats(orders):
    if not isinstance(orders, list):
        raise TypeError("Orders must be provided as a list.")

    valid_orders = []

    for order in orders:
        if (
            isinstance(order, (int, float))
            and not isinstance(order, bool)
            and order >= 0
        ):
            valid_orders.append(float(order))

    if not valid_orders:
        return {
            "count": 0,
            "average": None,
            "highest": None,
            "lowest": None
        }

    return {
        "count": len(valid_orders),
        "average": round(sum(valid_orders) / len(valid_orders), 2),
        "highest": max(valid_orders),
        "lowest": min(valid_orders)
    }


# Test 1
test_orders = [25.50, 40, "bad", -5, 100.25]
print("Test 1:")
print(calculate_order_stats(test_orders))

# Test 2
print("\nTest 2 - Empty List:")
print(calculate_order_stats([]))
```
